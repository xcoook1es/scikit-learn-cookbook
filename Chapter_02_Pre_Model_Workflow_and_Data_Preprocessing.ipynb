## Resep 1 — Menangani Data Hilang (Missing Data)

### Teori
Sebelum memilih metode imputasi, penting memahami **mekanisme** hilangnya data (Rubin, 1976):

- **MCAR** (*Missing Completely At Random*): peluang hilang tidak bergantung pada data apa pun. Contoh: sensor mati acak. Imputasi sederhana relatif aman.
- **MAR** (*Missing At Random*): peluang hilang bergantung pada variabel lain yang teramati. Contoh: responden muda lebih sering tidak mengisi pendapatan. Imputasi berbasis fitur lain (KNN/iteratif) lebih tepat.
- **MNAR** (*Missing Not At Random*): peluang hilang bergantung pada nilai yang hilang itu sendiri. Contoh: orang berpendapatan sangat tinggi enggan menyebutkan pendapatan. Tidak ada imputasi yang sepenuhnya bisa memperbaiki bias ini.

Tiga strategi di scikit-learn:

1. **`SimpleImputer`** — mengganti `NaN` di kolom $j$ dengan statistik kolom tersebut: mean $\bar x_j$, median, modus, atau konstanta. Cepat, tetapi **mengecilkan varians** fitur dan mengabaikan hubungan antar-fitur.
2. **`KNNImputer`** — untuk sampel dengan nilai hilang, cari $k$ tetangga terdekat (berdasarkan fitur yang tersedia, dengan jarak *nan-euclidean*) lalu ambil rata-ratanya:
$$\hat x_{ij} = \frac{1}{k}\sum_{l \in \mathcal{N}_k(i)} x_{lj}$$
Jarak *nan-euclidean* menghitung jarak hanya dari koordinat yang tersedia lalu menskalakannya: $d(\mathbf a,\mathbf b) = \sqrt{\frac{p}{p_{\text{ada}}}\sum_{\text{ada}}(a_j-b_j)^2}$.
3. **`IterativeImputer`** (gaya MICE) — setiap fitur dengan nilai hilang dimodelkan sebagai fungsi regresi dari fitur lain, $x_j \approx f_j(\mathbf{x}_{-j})$ (default `BayesianRidge`). Proses diulang beberapa putaran (*round-robin*) sampai stabil. Paling kaya informasi, tetapi paling lambat dan masih berstatus eksperimental.

### Persiapan data
Kita buat 20 baris × 10 fitur acak, lalu sekitar 20% nilai pada tiap kolom dihapus secara acak (mekanisme MCAR).

```python
# Load libraries
import numpy as np
import pandas as pd

# Create a larger sample dataset with missing values
np.random.seed(2024)  # For reproducibility
n_samples = 20
n_features = 10

# Generate random data
data = {
    f"Feature{i+1}": np.random.uniform(0, 100, n_samples) for i in range(n_features)
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Randomly introduce missing values (approximately 20% of the data)
for column in df.columns:
    mask = np.random.random(n_samples) < 0.2
    df.loc[mask, column] = np.nan

# Display the DataFrame with missing values
display(df)
```

### `SimpleImputer` (strategi mean)

```python
# Load libraries
from sklearn.impute import SimpleImputer

# Initialize the SimpleImputer and set the strategy to "mean," "median", or "most_frequent"
imputer = SimpleImputer(strategy="mean")

# Fit and transform the data
imputed_data = imputer.fit_transform(df)
imputed_df = pd.DataFrame(imputed_data, columns=df.columns)
imputed_df
```

### `KNNImputer` (k = 2)

```python
# Load libraries
from sklearn.impute import KNNImputer

# Initialize the KNNImputer
knn_imputer = KNNImputer(n_neighbors=2)

# Fit and transform the data using the previously defined DataFrame
knn_imputed_data = knn_imputer.fit_transform(df)
knn_imputed_df = pd.DataFrame(knn_imputed_data, columns=df.columns)
knn_imputed_df
```

### `IterativeImputer`

```python
# Load libraries
from sklearn.experimental import enable_iterative_imputer # Experimental feature requires loading
from sklearn.impute import IterativeImputer

# Initialize the IterativeImputer
iterative_imputer = IterativeImputer()

# Fit and transform the data using the previously defined DataFrame
iterative_imputed_data = iterative_imputer.fit_transform(df)
iterative_imputed_df = pd.DataFrame(iterative_imputed_data, columns=df.columns)
iterative_imputed_df
```

**Interpretasi dan perbandingan ketiga imputer.**

- **`SimpleImputer`** mengisi semua sel kosong dalam satu kolom dengan **nilai yang sama** (misalnya seluruh `NaN` di `Feature1` menjadi 52,56 = rata-rata kolom). Distribusi kolom jadi "menumpuk" di satu titik.
- **`KNNImputer`** menghasilkan nilai yang **berbeda-beda** per baris, karena setiap nilai diambil dari rata-rata 2 baris paling mirip.
- **`IterativeImputer`** di sini menghasilkan nilai yang sangat dekat dengan rata-rata kolom (misalnya 52,54 dan 52,63 untuk `Feature1`). Ini masuk akal: data dibangkitkan **acak dan saling independen**, sehingga model regresi tidak menemukan hubungan antar-fitur dan prediksinya mendekati rata-rata. Pada data nyata yang fiturnya saling berkorelasi, `IterativeImputer` dan `KNNImputer` biasanya jauh lebih informatif daripada `SimpleImputer`.

Pelajaran: metode imputasi canggih hanya unggul bila **ada struktur** (korelasi) antar-fitur yang bisa dimanfaatkan.

## Resep 2 — Teknik Penskalaan (Scaling)

### Teori
Algoritma berbasis **jarak** (KNN, K-Means, SVM RBF) dan berbasis **gradien** (regresi logistik, neural network) sensitif terhadap skala fitur. Jika satu fitur bernilai ribuan dan fitur lain bernilai 0–1, jarak Euclidean $\sqrt{\sum_j (a_j-b_j)^2}$ hampir seluruhnya ditentukan oleh fitur besar. Model berbasis pohon (Decision Tree, Random Forest) **tidak** terpengaruh skala karena hanya membandingkan nilai dengan ambang.

| Scaler | Rumus | Operasi per | Hasil |
|--------|-------|-------------|-------|
| `StandardScaler` | $z = \dfrac{x - \mu}{\sigma}$ | kolom (fitur) | mean 0, std 1; tetap sensitif outlier |
| `MinMaxScaler` | $x' = \dfrac{x - x_{\min}}{x_{\max} - x_{\min}}$ | kolom (fitur) | rentang [0, 1]; sangat sensitif outlier |
| `Normalizer` | $\mathbf{x}' = \dfrac{\mathbf{x}}{\lVert\mathbf{x}\rVert_2}$ | **baris (sampel)** | setiap sampel panjangnya 1 |

Perbedaan penting: `StandardScaler` dan `MinMaxScaler` bekerja **per fitur**, sedangkan `Normalizer` bekerja **per sampel**. `Normalizer` berguna bila yang penting adalah *arah* vektor, bukan besarnya — misalnya vektor frekuensi kata pada teks dengan *cosine similarity*.

```python
# Load libraries
from sklearn.preprocessing import StandardScaler

# Initialize the StandardScaler
scaler = StandardScaler()

# Fit and transform the data using the iterative imputed DataFrame
scaled_data = scaler.fit_transform(iterative_imputed_df)
scaled_df = pd.DataFrame(scaled_data, columns=iterative_imputed_df.columns)
scaled_df
```

```python
# Load libraries
from sklearn.preprocessing import MinMaxScaler

# Initialize the MinMaxScaler
minmax_scaler = MinMaxScaler()

# Fit and transform the data using the iterative imputed DataFrame
minmax_scaled_data = minmax_scaler.fit_transform(iterative_imputed_df)
minmax_scaled_df = pd.DataFrame(
    minmax_scaled_data, columns=iterative_imputed_df.columns
)
minmax_scaled_df
```

```python
# Load libraries
from sklearn.preprocessing import Normalizer

# Initialize the Normalizer
normalizer = Normalizer()

# Fit and transform the data using the iterative imputed DataFrame
normalized_data = normalizer.fit_transform(iterative_imputed_df)
normalized_df = pd.DataFrame(normalized_data, columns=iterative_imputed_df.columns)
normalized_df
```

```python
# Verifikasi sifat masing-masing scaler
print('StandardScaler -> mean kolom ~0 :', np.allclose(scaled_df.mean(), 0), '| std kolom ~1:', np.allclose(scaled_df.std(ddof=0), 1))
print('MinMaxScaler   -> min kolom     :', minmax_scaled_df.min().min(), '| max kolom:', minmax_scaled_df.max().max())
print('Normalizer     -> norma L2 tiap baris:', np.linalg.norm(normalized_df.values, axis=1).round(6)[:5], '...')
```

**Interpretasi.** Verifikasi di atas menunjukkan sifat yang dijanjikan rumus: kolom hasil `StandardScaler` ber-mean 0 dan ber-std 1, kolom hasil `MinMaxScaler` berada tepat di rentang [0, 1], dan setiap **baris** hasil `Normalizer` memiliki panjang (norma L2) 1.

## Resep 3 — Encoding Variabel Kategorikal

### Teori
Model matematis memerlukan input numerik. Cara mengubah kategori menjadi angka berpengaruh pada apa yang "dipercaya" model:

- **One-hot encoding**: kategori ke-$k$ dari $K$ kategori diubah menjadi vektor indikator $\mathbf{e}_k \in \{0,1\}^K$. Tidak ada urutan yang dipaksakan, dan jarak antar dua kategori berbeda selalu sama ($\sqrt 2$). Kekurangannya: dimensi bertambah sebanyak $K$ untuk tiap fitur. Pada model linear dengan intersep, $K$ kolom selalu berjumlah 1 sehingga terjadi multikolinearitas sempurna (*dummy variable trap*) — bisa diatasi dengan `drop='first'`.
- **Label/ordinal encoding**: kategori dipetakan ke $0, 1, \dots, K-1$. Ringkas, tetapi model linear/berbasis jarak akan menganggap ada urutan ($A < B < C$) dan jarak $|0-2| > |0-1|$ yang sebenarnya tidak ada. Di scikit-learn, `LabelEncoder` sebenarnya dimaksudkan untuk **target** $y$; untuk fitur gunakan `OrdinalEncoder`. Encoding urutan wajar dipakai untuk variabel yang memang ordinal (Junior < Senior < Manager) atau untuk model pohon.
- **`ColumnTransformer`**: menerapkan transformasi berbeda pada kolom berbeda (misal scaling untuk kolom numerik, one-hot untuk kolom kategorikal), lalu menggabungkan hasilnya secara horizontal.

### Persiapan data

```python
# Load libraries
import numpy as np

# Create sample categorical data with 20 records
np.random.seed(2024)  # for reproducibility
categories = ["A", "B", "C", "D"]
categorical_data = pd.DataFrame(
    {
        "Department": np.random.choice(categories, size=20),
        "Position": np.random.choice(["Junior", "Senior", "Manager"], size=20),
        "Location": np.random.choice(["NY", "SF", "LA", "CHI"], size=20),
    }
)

# Display the DataFrame with categorical values
display(categorical_data)
```

### `OneHotEncoder`

```python
# Load libraries
from sklearn.preprocessing import OneHotEncoder

# Initialize the OneHotEncoder
onehot_encoder = OneHotEncoder(sparse_output=False)

# Fit and transform the data
onehot_encoded_data = onehot_encoder.fit_transform(categorical_data)
onehot_encoded_df = pd.DataFrame(
    onehot_encoded_data, columns=onehot_encoder.get_feature_names_out()
)
onehot_encoded_df
```

### `LabelEncoder`

```python
# Load libraries
from sklearn.preprocessing import LabelEncoder

# Initialize the LabelEncoder
label_encoder = LabelEncoder()

# Create a new DataFrame to store label encoded values
label_encoded_df = pd.DataFrame()

# Fit and transform each categorical column
for column in categorical_data.columns:
    label_encoded_df[f"{column}_encoded"] = label_encoder.fit_transform(
        categorical_data[column]
    )
label_encoded_df
```

### `ColumnTransformer` untuk data campuran (numerik + kategorikal)

```python
# Load libraries
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler

# Create sample mixed data with 20 records
np.random.seed(2024)  # for reproducibility
mixed_data = pd.DataFrame(
    {
        "Age": np.random.randint(25, 65, size=20),
        "Salary": np.round(np.random.normal(60000, 15000, size=20), 2),
        "Experience": np.random.randint(1, 20, size=20),
        "Department": np.random.choice(["IT", "HR", "Sales", "Finance"], size=20),
        "Position": np.random.choice(["Junior", "Senior", "Manager"], size=20),
    }
)

# Display the DataFrame with mixed data
display(mixed_data)

# Initialize the ColumnTransformer
column_transformer = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), ["Age", "Salary", "Experience"]),
        ("cat", OneHotEncoder(), ["Department", "Position"]),
    ],
    remainder="passthrough",
)

# Fit and transform the data
transformed_data = column_transformer.fit_transform(mixed_data)

# Get feature names for the transformed columns
numeric_cols = ["Age_scaled", "Salary_scaled", "Experience_scaled"]
categorical_cols = column_transformer.named_transformers_["cat"].get_feature_names_out(
    ["Department", "Position"]
)

# Create the transformed DataFrame
transformed_df = pd.DataFrame(
    transformed_data, columns=numeric_cols + list(categorical_cols)
)
transformed_df
```

**Interpretasi.** Tiga kolom kategorikal dengan masing-masing 4, 3, dan 4 kategori menghasilkan $4+3+4 = 11$ kolom one-hot, dan setiap baris berisi tepat tiga angka 1. Pada `ColumnTransformer`, 3 kolom numerik distandardisasi dan 2 kolom kategorikal (4 dan 3 kategori) menjadi 7 kolom one-hot, sehingga total 10 kolom.

## Resep 4 — Pengantar Pipeline

### Teori
Pipeline menjamin bahwa urutan `fit_transform` pada data latih dan `transform` pada data uji selalu benar. Tanpa pipeline, kesalahan klasik adalah melakukan scaling pada **seluruh** dataset sebelum split; statistik $\mu,\sigma$ jadi ikut "melihat" data uji (**data leakage**) sehingga skor evaluasi terlalu optimistis.

Catatan untuk kode buku di bawah: kolom terakhir dari `transformed_df` (yaitu `Position_Senior`) dipakai sebagai target `y` hanya untuk keperluan demonstrasi mekanisme pipeline, bukan masalah prediksi yang bermakna.

```python
# Load libraries
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# First, separate features and target (assuming last column is target)
X = transformed_df.iloc[:, :-1]  # all columns except last
y = transformed_df.iloc[:, -1]  # last column

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=2024
)

# Create a pipeline
pipeline = Pipeline(
    [
        ("imputer", SimpleImputer(strategy="mean")),  # handle missing values
        ("scaler", StandardScaler()),  # scale the features
    ]
)

# Fit and transform the training data
X_train_transformed = pipeline.fit_transform(X_train)

# Transform the test data
X_test_transformed = pipeline.transform(X_test)

# Create DataFrames with the transformed data (to preserve column names)
X_train_transformed = pd.DataFrame(
    X_train_transformed, columns=X_train.columns, index=X_train.index
)

X_test_transformed = pd.DataFrame(
    X_test_transformed, columns=X_test.columns, index=X_test.index
)
```

```python
print('Bentuk X_train:', X_train_transformed.shape, '| X_test:', X_test_transformed.shape)
print('Mean kolom X_train (harus ~0):', X_train_transformed.mean().abs().max().round(6))
print('Mean kolom X_test (tidak harus 0):', X_test_transformed.mean().round(2).values[:4], '...')
```

## Resep 5 — Feature Engineering dan Seleksi Fitur

### Teori

**`PolynomialFeatures`** membuat semua monomial hingga derajat $d$. Untuk dua fitur $(x_1, x_2)$ dan $d=2$:
$$[1,\; x_1,\; x_2,\; x_1^2,\; x_1x_2,\; x_2^2]$$
Ini memungkinkan model linear menangkap hubungan non-linear dan interaksi. Banyaknya fitur keluaran adalah $\binom{p+d}{d}$, jadi tumbuh sangat cepat: untuk $p = 9$ dan $d = 2$ menjadi $\binom{11}{2} = 55$ kolom.

**`KBinsDiscretizer`** mengubah fitur kontinu menjadi $k$ interval (*bin*). Strategi `uniform` membuat bin dengan lebar sama, $\text{lebar} = (x_{\max} - x_{\min})/k$; `quantile` membuat bin dengan jumlah sampel sama; `kmeans` memakai K-Means 1D. Diskretisasi bisa membantu model linear menangkap pola non-linear berbentuk "tangga", tetapi membuang informasi detail.

**Seleksi fitur** mengurangi fitur yang tidak relevan → model lebih sederhana, lebih cepat, lebih mudah diinterpretasikan, dan kurang rawan overfitting:
- **RFE (Recursive Feature Elimination)**: latih model → buang fitur dengan bobot $|w_j|$ terkecil → ulangi sampai tersisa `n_features_to_select`. Atribut `ranking_` = 1 untuk fitur yang terpilih; angka makin besar berarti makin awal dibuang.
- **`SelectFromModel`**: latih model sekali, lalu pertahankan fitur yang "kepentingannya" (|koefisien| atau `feature_importances_`) di atas ambang (`threshold='mean'` → di atas rata-rata).

Catatan: koefisien model linear hanya bisa dibandingkan sebagai ukuran kepentingan bila fitur sudah berada pada skala yang sama — di sini terpenuhi karena data sudah distandardisasi oleh pipeline.

```python
# Load libraries
from sklearn.preprocessing import PolynomialFeatures

# Initialize the PolynomialFeatures
poly = PolynomialFeatures(degree=2)

# Fit and transform the X_train_transformed data
poly_features = poly.fit_transform(X_train_transformed)
poly_features_df = pd.DataFrame(
    poly_features, columns=poly.get_feature_names_out(X_train_transformed.columns)
)
poly_features_df
```

```python
# Load libraries
from sklearn.preprocessing import KBinsDiscretizer

# Initialize the KBinsDiscretizer
kbins = KBinsDiscretizer(n_bins=3, encode="ordinal", strategy="uniform")

# Fit and transform the X_train_transformed data
binned_data = kbins.fit_transform(X_train_transformed)
binned_df = pd.DataFrame(binned_data, columns=X_train_transformed.columns)
binned_df
```

```python
# Load libraries
from sklearn.feature_selection import RFE
from sklearn.linear_model import LinearRegression

# Initialize the RFE
rfe = RFE(estimator=LinearRegression(), n_features_to_select=1)

# Fit the RFE to the X_train_transformed and y_train data
rfe.fit(X_train_transformed, y_train)

# Get the ranking of features
rfe.ranking_
```

```python
# Load libraries
from sklearn.feature_selection import SelectFromModel
from sklearn.linear_model import LinearRegression

# Initialize SelectFromModel with LinearRegression
selector = SelectFromModel(
    estimator=LinearRegression(),
    prefit=False,
    threshold="mean",  # Use mean of feature importances as threshold
)

# Fit the selector
selector.fit(X_train_transformed, y_train)

# Get selected features
selected_features_mask = selector.get_support()

# Get feature names that were selected
selected_features = X_train_transformed.columns[selected_features_mask].tolist()

# Print feature importance scores and selection status
feature_importance = pd.DataFrame(
    {
        "Feature": X_train_transformed.columns,
        "Importance": selector.estimator_.coef_,
        "Selected": selected_features_mask,
    }
)
feature_importance.sort_values("Importance", key=abs, ascending=False)
```

**Interpretasi.**

- `PolynomialFeatures(degree=2)` pada 9 fitur menghasilkan $\binom{11}{2} = 55$ kolom: 1 bias, 9 linear, 9 kuadrat, dan 36 interaksi.
- `KBinsDiscretizer` mengubah setiap kolom menjadi kode bin 0, 1, atau 2. Untuk kolom one-hot (hanya bernilai dua macam setelah scaling), hasilnya hanya ujung-ujung bin.
- `RFE` memberi ranking 1 untuk `Position_Manager` dan ranking 2 untuk `Position_Junior`, sedangkan `SelectFromModel` hanya memilih dua fitur `Position_*` dengan koefisien ≈ −0,50 dan −0,43, sementara koefisien fitur lain ≈ 0 (orde $10^{-16}$, sisa pembulatan numerik).

Mengapa? Target pada contoh ini adalah kolom `Position_Senior`, dan untuk setiap baris berlaku $\text{Junior} + \text{Senior} + \text{Manager} = 1$. Artinya target **dapat dihitung tepat** dari dua kolom posisi lainnya, sehingga seleksi fitur dengan benar menemukan bahwa hanya dua kolom itu yang dibutuhkan. Ini sekaligus contoh nyata **target leakage**: jika suatu fitur merupakan turunan langsung dari target, model akan tampak sempurna padahal tidak belajar pola yang berguna. Pada proyek nyata, kasus seperti ini harus dicurigai dan diperiksa.

## Latihan Praktik — Pipeline Lengkap pada California Housing

Dataset California Housing (sensus 1990) berisi 20.640 blok wilayah dengan 8 fitur numerik (pendapatan median, umur rumah, rata-rata jumlah kamar, populasi, lintang/bujur, dll.). Targetnya adalah nilai median rumah dalam satuan \$100.000.

Pipeline: imputasi median → standardisasi → `RandomForestRegressor`. Skor yang dikembalikan `score()` adalah $R^2$ pada data uji.

> **Catatan:** `fetch_california_housing()` mengunduh dataset dari internet saat pertama kali dijalankan. Jalankan sel ini di Jupyter/Colab yang terhubung internet. Pada percobaan dengan konfigurasi ini, Random Forest biasanya memberi $R^2$ sekitar 0,80 pada data uji.

```python
# Load libraries
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor

# Load the California Housing dataset
california_data = fetch_california_housing()
X, y = california_data.data, california_data.target

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=2024
)

# Create a comprehensive pipeline
comprehensive_pipeline = Pipeline(
    [
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("model", RandomForestRegressor()),
    ]
)

# Fit the pipeline
comprehensive_pipeline.fit(X_train, y_train)

# Evaluate the pipeline
score = comprehensive_pipeline.score(X_test, y_test)
score
```

## Kesimpulan Bab 2

- Pilih metode imputasi sesuai mekanisme data hilang: `SimpleImputer` cepat tetapi mengabaikan hubungan antar-fitur; `KNNImputer` dan `IterativeImputer` memanfaatkan fitur lain.
- Scaling penting untuk model berbasis jarak/gradien. `StandardScaler` dan `MinMaxScaler` bekerja per fitur, `Normalizer` bekerja per sampel.
- One-hot untuk kategori nominal, ordinal encoding untuk kategori berurutan; `LabelEncoder` ditujukan untuk target. `ColumnTransformer` menangani data campuran.
- **Pipeline** merangkai semua langkah dan mencegah *data leakage* karena statistik preprocessing hanya dipelajari dari data latih.
- `PolynomialFeatures` dan `KBinsDiscretizer` menambah kemampuan model linear menangkap non-linearitas; `RFE` dan `SelectFromModel` mengurangi fitur yang tidak penting.
